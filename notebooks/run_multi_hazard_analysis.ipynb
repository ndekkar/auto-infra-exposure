This notebook allows you to **run the full multi-hazard infrastructure exposure analysis in one single execution**.

- select which hazards to analyse,

## What this notebook does

When executed, this notebook will:

1. Read the project configuration file (`config.yaml`)
2. Load all required input data (AOI, infrastructure layers, hazard datasets)
3. Automatically run exposure analyses for all **activated hazards**
4. Generate:

   - spatial output files,
   - summary results for each hazard

## How to use this notebook
